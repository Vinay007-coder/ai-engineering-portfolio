# Tensor Indexing & Slicing

```python
# Creating a 3D tensor
tensor_3d = np.array([
    [ # Page 0
        [10, 11, 12],
        [13, 14, 15],
        [16, 17, 18]
    ],
    [ # Page 1
        [20, 21, 22],
        [23, 24, 25],
        [26, 27, 28]
    ]
])
```

```python
# Tensor Indexing (Grabbing a Single Number)
# Goal: Grab the number 24 from Page 1, Row 1, Column 1
single_number = tensor_3d[1, 1, 1]
print("Indexed Number:", single_number)
# Output: 24

# Goal: Grab the number 12 from Page 0, Row 0, Column 2
another_number = tensor_3d[0, 0, 2]
print("Indexed Number:", another_number)
# Output: 12
```

```python
# Tensor Slicing (Extracting Sections)

# Slice a single page
# Grab everything on Page 0 (leaving row and column blank means get all of them)
page_0 = tensor_3d[0, :, :]
print("Page 0 Matrix:\n", page_0)
# Output:
# [[10 11 12]
#  [13 14 15]
#  [16 17 18]]
```

```python
# Slice a specific row across all pages
# Grab Row 0 from BOTH pages
rows_0 = tensor_3d[:, 0, :]
print("Row 0 from all pages:\n", rows_0)
# Output:
# [[10 11 12]
#  [20 21 22]]
```

```python
# Slice a specific sub-grid (The center square)
# From Page 1, grab rows 0 to 2 and columns 0 to 2
# Remember: '0:2' means it grabs index 0 and 1, but stops before 2.
sub_grid = tensor_3d[1, 0:2, 0:2]
print("Sub-grid:\n", sub_grid)
# Output:
# [[20 21]
#  [23 24]]
```

# **Matrix multiplication**
Two main ways of performing multiplication in neural networks and deep learning:
1. Element-wise multiplication
2. Matrix multiplication (dot product)

There are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:

*   (3, 2) @ (3, 2) won't work
*   (2, 3) @ (3, 2) will work
*   (3, 2) @ (2, 3) will work

2. The resulting matrix has the shape of the outer dimensions:

*   (2, 3) @ (3, 2) -> (2, 2)
*   (3, 2) @ (2, 3) -> (3, 3)







# Dealing with tensor shape errors

```python
tensor_A = torch.tensor([[1,2], [3,4], [5,6]])
tensor_A
```

```python
tensor_B = torch.tensor([[2,3], [3,4], [5,6]])
tensor_B
```

```python
# torch.mm(tensor_A, tensor_B) -> Alias for the torch.matmul()
torch.matmul(tensor_A, tensor_B)
# Output: RuntimeError: mat1 and mat2 shapes cannot be multiplied (3x2 and 3x2)
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose**.

A **transpose** switches the axes or dimensions of a given tensor.

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
# Satisfies our first rule: (3,2) (2,3)
```

```python
output = torch.matmul(tensor_A, tensor_B.T)
# We will get a successful output
```

```python
output.shape
# Output: torch.Size([3, 3]) -> Satisfies our second rule
```

# **Reshaping, stacking, squeezing, and unsqueezing tensors**

*   **Reshaping** - reshapes an input tensor to a defined shape

*   **View** - Return a view of an input tensor of certain shape but keep the same memory as the original tensor

*   **Stacking** - combine multiple tensors on top of each other (vstack) or side by side (hstack)

*   **Squeeze** - removes all 1 dimensions from a tensor

*   **Unsqueeze** - add a 1 dimension to a target tensor

*   **Permute** - Return a view of the input with dimensions permuted (swapped) in a certain way







```python
# Let's create a tensor
x = torch.arange(1., 10.)
x, x.shape, x.ndim
```

```python
# Let's Add an extra dimension
# x_reshaped = x.reshape(1,7)
# When we tried to put a tensor of size 7 into 9, we got this error:
# RuntimeError: shape '[1, 7]' is invalid for input of size 9
# x_reshaped = x.reshape(2,9)
# We can't reshape this tensor by trying to make it 2 dimensional, because the size will be 2 * 9, which is 18 and that's more than our input size 9
x_reshaped = x.reshape(1,9)
x_reshaped, x_reshaped.shape, x_reshaped.ndim
```

```python
x = torch.arange(1., 11.)
x, x.shape, x.ndim
```

```python
#  Add an extra dimension
x_reshaped = x.reshape(5, 2)
x_reshaped, x_reshaped.shape, x_reshaped.ndim
```

```python
# Change the view
z = x.view(1, 10)
z, z.shape
```

| Code         | What it extracts                                 |
| ------------ | ------------------------------------------------ |
| `z[2, 3]`    | A single number at Row 2, Column 3               |
| `z[1, :]`    | The entire second row                            |
| `z[:, -1]`   | The entire last column                           |
| `z[0:3, :]`  | The first 3 rows                                 |
| `z[:, ::-1]` | Flips the matrix horizontally (reverses columns) |


```python
# Changing z changes x, because a view of a tensor shares the same memory as the original input
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked, x_stacked.shape, x_stacked.ndim
```

```python
# Creating two 1D tensors with 3 elements each
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
a, b
```

```python
# Vertical Stacking (torch.vstack)
# Stack vertically
v_result = torch.vstack([a, b])

print("Vertical Stack Result:\n", v_result)
print("Resulting Shape:", v_result.shape)

# Output Matrix (2 rows, 3 columns):
# [[1, 2, 3],
#  [4, 5, 6]]
# Shape: torch.Size([2, 3])
```

```python
# Horizontal Stacking (torch.hstack)
# Stack horizontally
h_result = torch.hstack([a, b])

print("Horizontal Stack Result:", h_result)
print("Resulting Shape:", h_result.shape)

# Output Vector (1 row with 6 elements):
# [1, 2, 3, 4, 5, 6]
# Shape: torch.Size([6])
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - adds a single dimension to a target tensor at a specific dim (dimension)
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
#  torch.permute - rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224, 224, 3)) # [height, width, colour_channels]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}") # [colour_channels, height, width]
```

```python
x_original[0,0,0] = 0.9000
x_original[0,0,0], x_permuted[0,0,0]
# Returns a view of the original tensor input with its dimensions permuted.
```

# PyTorch tensors & NumPy
NumPy is a popular scientific Python numerical computing library.
And because of this, PyTorch has functionality to interact with it.

*   Data in NumPy -> PyTorch tensor -> torch.from_numpy(ndarray)
*   PyTorch tensor -> NumPy -> torch.Tensor.numpy()



```python
# Numpy array to tensor
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
# When converting from numpy -> pytorch, pytorch reflects numpy's default datatype of float64 unless specified otherwise.
array, tensor
```

```python
# Change the value of array, what will this do to `tensor`?
array = array + 1
array, tensor
# They don't share same memory, so nothing will happen to `tensor`
```

```python
# Tensor to NumPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
# When converting from tensor -> numpy, numpy reflects pytorch's default datatype of float32 unless specified otherwise
tensor, numpy_tensor
```

```python
# Change the tensor, what happens to `numpy_tensor`?
tensor = tensor + 1
tensor, numpy_tensor
# They don't share same memory, so nothing will happen to `numpy_tensor`
```

# Reproducibility (trying to take random out of random)
In short how a neural network learns:

`start with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> again -> again -> again...`

To reduce the randomness in neural networks and PyTorch, here comes the concept of a **random seed**.

Essentially what the random seed does is "flavour" the randomness.

```python
# create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}\n")
print(f"Tensor B:\n{random_tensor_B}\n")
print(f"Does Tensor A equal Tensor B? (anywhere)")
random_tensor_A == random_tensor_B
```

```python
# Let's make some random but reproducible tensors

# Set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")
print(f"Does Tensor C equal Tensor D? (anywhere)")
random_tensor_C == random_tensor_D
```
